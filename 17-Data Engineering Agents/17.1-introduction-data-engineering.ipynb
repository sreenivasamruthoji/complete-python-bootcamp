# Introduction to Data Engineering

Data Engineering is the practice of designing and building systems for collecting, storing, and analyzing data at scale. Data engineers create the infrastructure and tools that enable data scientists and analysts to work effectively.

## What is Data Engineering?

Data engineering involves:
- **Data Collection**: Gathering data from various sources
- **Data Storage**: Organizing and storing data efficiently
- **Data Processing**: Transforming raw data into usable formats
- **Data Pipeline**: Automating the flow of data from source to destination
- **Data Quality**: Ensuring data accuracy and reliability

## Key Concepts in Data Engineering

### 1. ETL vs ELT
- **ETL (Extract, Transform, Load)**: Transform data before loading
- **ELT (Extract, Load, Transform)**: Load data first, then transform

### 2. Batch vs Stream Processing
- **Batch Processing**: Process large volumes of data at scheduled intervals
- **Stream Processing**: Process data in real-time as it arrives

### 3. Data Warehouse vs Data Lake
- **Data Warehouse**: Structured, processed data for analytics
- **Data Lake**: Raw, unprocessed data in various formats

## Simple Data Pipeline Example

Let's create a basic data pipeline that:
1. Extracts data from a source
2. Transforms the data
3. Loads it to a destination

```python
import pandas as pd
from datetime import datetime
import json

# Sample data source (simulating API response)
raw_data = {
    'users': [
        {'id': 1, 'name': 'Alice', 'age': 28, 'country': 'USA'},
        {'id': 2, 'name': 'Bob', 'age': 35, 'country': 'UK'},
        {'id': 3, 'name': 'Charlie', 'age': 42, 'country': 'Canada'},
        {'id': 4, 'name': 'Diana', 'age': 31, 'country': 'USA'}
    ]
}

print("Raw data:")
print(json.dumps(raw_data, indent=2))
```

```python
# Extract: Get data from source
def extract_data(source):
    """Extract data from the source"""
    return source['users']

users = extract_data(raw_data)
print(f"Extracted {len(users)} users")
```

```python
# Transform: Clean and enrich the data
def transform_data(users):
    """Transform the extracted data"""
    df = pd.DataFrame(users)
    
    # Add a timestamp
    df['processed_at'] = datetime.now().isoformat()
    
    # Add age group categorization
    df['age_group'] = pd.cut(df['age'], 
                             bins=[0, 30, 40, 100], 
                             labels=['Young', 'Middle', 'Senior'])
    
    # Convert to uppercase for standardization
    df['country'] = df['country'].str.upper()
    
    return df

transformed_df = transform_data(users)
print("\nTransformed data:")
print(transformed_df)
```

```python
# Load: Save the transformed data
def load_data(df, destination):
    """Load transformed data to destination"""
    df.to_csv(destination, index=False)
    print(f"\nData loaded to {destination}")
    return True

# Save to CSV file
load_data(transformed_df, 'processed_users.csv')
```

## Data Pipeline Orchestration

A complete pipeline combines all steps:

```python
def run_pipeline(source_data, destination_file):
    """Run the complete ETL pipeline"""
    try:
        # Extract
        print("Step 1: Extracting data...")
        data = extract_data(source_data)
        
        # Transform
        print("Step 2: Transforming data...")
        transformed = transform_data(data)
        
        # Load
        print("Step 3: Loading data...")
        load_data(transformed, destination_file)
        
        print("\n✓ Pipeline completed successfully!")
        return True
    except Exception as e:
        print(f"✗ Pipeline failed: {str(e)}")
        return False

# Run the pipeline
run_pipeline(raw_data, 'final_users.csv')
```

## Best Practices

1. **Idempotency**: Pipelines should produce the same result when run multiple times
2. **Error Handling**: Always handle and log errors gracefully
3. **Data Validation**: Validate data at each step
4. **Monitoring**: Track pipeline performance and failures
5. **Documentation**: Document data sources, transformations, and schemas
6. **Testing**: Test pipelines with sample data before production
